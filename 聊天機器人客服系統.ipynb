```python
# ==========================================
# Python 聊天機器人客服系統
# ==========================================

# 儲存常見問題與客服回答
faq = {
    "訂單": "您好，您可以提供訂單編號，我們可以協助您查詢訂單狀態。",
    "退貨": "您好，商品收到後 7 天內可以申請退貨。",
    "付款": "我們目前提供信用卡、ATM 轉帳與超商付款。",
    "營業時間": "客服服務時間為週一至週五 09:00～18:00。",
    "運費": "單筆訂單滿 1,000 元即可享有免運優惠。"
}


# 顯示歡迎訊息
print("================================")
print("      歡迎使用線上客服機器人")
print("================================")
print("您好！我是 AI 客服機器人，很高興為您服務。")
print("您可以詢問：訂單、退貨、付款、營業時間、運費")
print("輸入「離開」即可結束服務。")
print()


# 開始客服對話
while True:

    # 讓使用者輸入問題
    user_input = input("請輸入您的問題：")

    # 如果使用者輸入「離開」，結束程式
    if user_input == "離開":
        print("感謝您的使用，祝您有美好的一天！")
        break

    # 記錄是否找到對應的問題
    found = False

    # 檢查使用者的問題中是否包含 FAQ 關鍵字
    for keyword in faq:

        if keyword in user_input:
            print("客服：", faq[keyword])
            found = True
            break

    # 如果沒有找到對應的問題
    if not found:
        print("客服：抱歉，我目前無法理解您的問題。")
        print("客服：您可以詢問「訂單」、「退貨」、「付款」、「營業時間」或「運費」。")

    print()
```
