```python
# 用 list 儲存所有記帳資料
# 每一筆帳目會使用 dictionary 儲存
records = []


# 使用 while True 建立一個「持續執行」的選單
# 只要使用者沒有選擇離開，系統就會一直執行
while True:

    # 顯示主選單
    print("\n===== 個人記帳系統 =====")
    print("1. 新增收入")
    print("2. 新增支出")
    print("3. 查看所有帳目")
    print("4. 查看目前餘額")
    print("5. 離開")

    # 使用 input() 讓使用者輸入選項
    # input() 取得的資料預設是字串（str）
    choice = input("請選擇功能：")


    # ==============================
    # 1. 新增收入
    # ==============================
    if choice == "1":

        # input() 取得金額後，使用 float()
        # 將字串轉換成浮點數
        #
        # 例如：
        # input() 得到 "30000"
        # float("30000") → 30000.0
        amount = float(input("請輸入收入金額："))

        # 輸入收入類別
        # 例如：薪水、獎學金、兼職收入
        category = input("請輸入收入類別：")

        # 輸入這筆收入的備註
        # 例如：9 月薪水
        description = input("請輸入備註：")


        # 將這筆收入加入 records
        #
        # append() 可以將一個新的元素加入 list 的最後面
        #
        # 每一筆帳目使用 dictionary 儲存，
        # 包含：
        # type        → 帳目的類型
        # amount      → 金額
        # category    → 類別
        # description → 備註
        records.append({
            "type": "收入",
            "amount": amount,
            "category": category,
            "description": description
        })

        # 顯示新增成功訊息
        print("收入新增成功！")


    # ==============================
    # 2. 新增支出
    # ==============================
    elif choice == "2":

        # 輸入支出金額
        # 使用 float() 將 input() 得到的字串轉成數字
        amount = float(input("請輸入支出金額："))

        # 輸入支出類別
        # 例如：飲食、交通、娛樂、購物
        category = input("請輸入支出類別：")

        # 輸入支出備註
        # 例如：午餐、捷運、電影
        description = input("請輸入備註：")


        # 將這筆支出加入 records
        records.append({
            "type": "支出",
            "amount": amount,
            "category": category,
            "description": description
        })

        # 顯示新增成功訊息
        print("支出新增成功！")


    # ==============================
    # 3. 查看所有帳目
    # ==============================
    elif choice == "3":

        print("\n===== 所有帳目 =====")


        # len(records) 可以取得 records 裡面
        # 總共有幾筆資料
        #
        # 如果長度是 0，代表目前沒有任何帳目
        if len(records) == 0:

            print("目前沒有任何帳目。")


        # 如果 records 裡面有資料
        else:

            # 使用 for 迴圈逐筆查看 records
            #
            # enumerate() 可以同時取得：
            # i      → 編號
            # record → 每一筆帳目
            #
            # start=1 代表編號從 1 開始
            for i, record in enumerate(records, start=1):

                # 顯示每一筆帳目的資訊
                #
                # record["type"]
                # → 取得帳目類型
                #
                # record["amount"]
                # → 取得金額
                #
                # :.2f
                # → 將數字顯示到小數點後 2 位
                #
                # record["category"]
                # → 取得類別
                #
                # record["description"]
                # → 取得備註
                print(
                    f"{i}. "
                    f"{record['type']} | "
                    f"${record['amount']:.2f} | "
                    f"{record['category']} | "
                    f"{record['description']}"
                )


    # ==============================
    # 4. 查看目前餘額
    # ==============================
    elif choice == "4":

        # 建立總收入變數
        # 初始值設定為 0
        income = 0

        # 建立總支出變數
        # 初始值設定為 0
        expense = 0


        # 逐筆查看所有帳目
        for record in records:

            # 如果這筆帳目的類型是「收入」
            if record["type"] == "收入":

                # 將這筆收入金額累加到 income
                #
                # 例如：
                # income = 30000
                # 下一筆收入 = 5000
                #
                # income += 5000
                # → income = 35000
                income += record["amount"]


            # 如果不是收入，就是支出
            else:

                # 將支出金額累加到 expense
                expense += record["amount"]


        # 計算目前餘額
        #
        # 餘額 = 總收入 - 總支出
        balance = income - expense


        # 顯示統計結果
        print(f"\n總收入：${income:.2f}")
        print(f"總支出：${expense:.2f}")
        print(f"目前餘額：${balance:.2f}")


    # ==============================
    # 5. 離開系統
    # ==============================
    elif choice == "5":

        # 顯示離開訊息
        print("感謝使用個人記帳系統！")

        # break 可以立即結束 while 迴圈
        # 因此程式也會結束
        break


    # ==============================
    # 使用者輸入不存在的選項
    # ==============================
    else:

        # 如果使用者輸入 1～5 以外的內容
        # 顯示錯誤訊息
        print("無效的選擇，請重新輸入。")
```
