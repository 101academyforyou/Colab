```python
import pandas as pd


# 用 list 儲存所有記帳資料
# 每一筆帳目會使用 dictionary 儲存
records = []


# 持續執行記帳系統
while True:

    # 顯示主選單
    print("\n===== 個人記帳系統 =====")
    print("1. 新增收入")
    print("2. 新增支出")
    print("3. 查看所有帳目")
    print("4. 查看目前餘額")
    print("5. 離開")
    print("6. 儲存到 Excel")

    # 讓使用者選擇功能
    choice = input("請選擇功能：")


    # ==============================
    # 1. 新增收入
    # ==============================
    if choice == "1":

        # 輸入收入金額
        # float() 將文字轉換成浮點數
        amount = float(input("請輸入收入金額："))

        # 輸入收入類別
        category = input("請輸入收入類別：")

        # 輸入備註
        description = input("請輸入備註：")

        # 將這筆收入加入 records
        records.append({
            "type": "收入",
            "amount": amount,
            "category": category,
            "description": description
        })

        print("收入新增成功！")


    # ==============================
    # 2. 新增支出
    # ==============================
    elif choice == "2":

        # 輸入支出金額
        amount = float(input("請輸入支出金額："))

        # 輸入支出類別
        category = input("請輸入支出類別：")

        # 輸入備註
        description = input("請輸入備註：")

        # 將這筆支出加入 records
        records.append({
            "type": "支出",
            "amount": amount,
            "category": category,
            "description": description
        })

        print("支出新增成功！")


    # ==============================
    # 3. 查看所有帳目
    # ==============================
    elif choice == "3":

        print("\n===== 所有帳目 =====")

        # 判斷目前是否有帳目
        if len(records) == 0:

            print("目前沒有任何帳目。")

        else:

            # 使用 for 逐筆顯示帳目
            for i, record in enumerate(records, start=1):

                print(
                    f"{i}. "
                    f"{record['type']} | "
                    f"${record['amount']:.2f} | "
                    f"{record['category']} | "
                    f"{record['description']}"
                )


    # ==============================
    # 4. 查看目前餘額
    # ==============================
    elif choice == "4":

        # 初始化總收入與總支出
        income = 0
        expense = 0

        # 逐筆計算
        for record in records:

            # 如果是收入
            if record["type"] == "收入":
                income += record["amount"]

            # 如果是支出
            else:
                expense += record["amount"]

        # 計算餘額
        balance = income - expense

        # 顯示結果
        print(f"\n總收入：${income:.2f}")
        print(f"總支出：${expense:.2f}")
        print(f"目前餘額：${balance:.2f}")


    # ==============================
    # 5. 離開系統
    # ==============================
    elif choice == "5":

        print("感謝使用個人記帳系統！")

        # 結束 while 迴圈
        break


    # ==============================
    # 6. 儲存到 Excel
    # ==============================
    elif choice == "6":

        # 確認目前是否有任何帳目
        if len(records) == 0:

            print("目前沒有任何帳目，無法儲存。")

        else:

            # 將 records 轉換成 Pandas DataFrame
            #
            # records 原本是：
            # [
            #     {"type": "收入", "amount": 30000, ...},
            #     {"type": "支出", "amount": 150, ...}
            # ]
            #
            # DataFrame 可以將這些資料整理成表格
            df = pd.DataFrame(records)

            # 設定 Excel 檔案名稱
            file_name = "/content/personal_accounting.xlsx"

            # 將 DataFrame 儲存成 Excel
            #
            # index=False：
            # 不要將 Pandas 自動產生的索引存進 Excel
            df.to_excel(file_name, index=False)

            print(f"帳目已儲存到：{file_name}")


    # ==============================
    # 無效選項
    # ==============================
    else:

        print("無效的選擇，請重新輸入。")
```
